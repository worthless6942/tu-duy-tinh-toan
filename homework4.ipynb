W4A20 Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không 

```python
n = int(input("Nhap n: "))

if n <= 0:
    print("Khong phai 2^k")
else:
    while n % 2 == 0:
        n //= 2

    if n == 1:
        print("La 2^k")
    else:
        print("Khong phai 2^k")
```

W4A22 Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ

```python
n = input("Nhap n:")
chan = 0
le = 0
for i in n:
    if i.isdigit():
        if int(i)%2 == 0:
            chan += 1
        else:
            le += 1

print("So chu so chan la: ", chan)
print("So chu so le la: ", le)
```

W4A24
Nhập vào A, tìm n nhỏ nhất sao cho: 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A

```python
A = float(input("Nhập A: "))

tong = 0.0
n = 0

while tong <= A:
    n += 1
    tong += 1.0 / n

print("n nhỏ nhất là:", n)
```

W4A25
Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1.
Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập 


```python
lst = []

tt = True

while tt:
    n = int(input("Nhap n vao chuoi: "))
    if n == -1:
        tt = False
    else:
        lst.append(n)

print("So lon nhat la: ", max(lst))
print("So nho nhat la: ", min(lst))

```

Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ
bằng tổng hai số trước đó
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt
quá A

```python
lst = [1, 1]
a = int(input("Nhap a:"))
f1, f2 = 1, 1
while f1 <= a:
    lst.append(f1)
    f1, f2 =  f2, f1 + f2
print("So can tim la:", max(lst))
    
```

W4A27
Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu từ (quy định là chuỗi không có
ký tự đặc biệt, không số, không có dấu câu, chỉ có ký tự chữ và khoảng trắng)
Dùng cách: for trên index; KHÔNG dùng hàm split(), len()

```python
s = input("Nhập chuỗi: ")


n = 0
for _ in s:
    n += 1


count = 0

for i in range(n):
    if s[i] != ' ' and (i == 0 or s[i-1] == ' '):
        count += 1

print("Số từ là:", count)
```

W4A29: Cho một chuỗi chỉ gồm các ký tự chữ cái thường và chữ cái hoa, hãy đổi ký tự hoa thành
thường và thường thành hoa. 


```python
n = input("Nhap chuoi: ")
lst = []
for i in n:
    if i.isupper():
        lst.append(i.lower())
    if i.islower():
        lst.append(i.upper())

print("Chuoi la: ", *lst,sep = '')
```

W4A31: 
Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in thường, bao nhiêu ký tự số

```python
s = input("Nhập chuỗi: ")

upper = 0
lower = 0
digit = 0

for ch in s:
    if 'A' <= ch <= 'Z':
        upper += 1
    elif 'a' <= ch <= 'z':
        lower += 1
    elif '0' <= ch <= '9':
        digit += 1

print("Chữ in hoa:", upper)
print("Chữ thường:", lower)
print("Chữ số:", digit)
```

bai 14

```python
x = int(input("Nhập số (-1 để dừng): "))

# Kiểm tra nếu nhập -1 ngay từ đầu
if x == -1:
    print("Không có dữ liệu")
else:
    max_val = x
    min_val = x

    while True:
        x = int(input("Nhập số (-1 để dừng): "))
        if x == -1:
            break

        if x > max_val:
            max_val = x
        if x < min_val:
            min_val = x

    print("Số lớn nhất:", max_val)
    print("Số nhỏ nhất:", min_val)
```

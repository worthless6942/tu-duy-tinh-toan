W7A16
Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
a = list(map(int, input().split()))

d = {}

for i in a:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1

xoa = 0

for x in d:
    if d[x] < x:
        xoa += d[x]
    elif d[x] > x:
        xoa += d[x] - x

if xoa == 0:
    print("Mang dep")
else:
    print(xoa)
```

W7A17
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. 

```python
a = list(map(int, input().split()))

cand = None
count = 0

for x in a:
    if count == 0:
        cand = x
        count = 1
    elif x == cand:
        count += 1
    else:
        count -= 1

if a.count(cand) > len(a) // 2:
    print(cand)
```

Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng dần, lẻ giảm dần; ;

W7A19
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không dùng bất cứ thuật toán sắp xếp nào

```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))

if len(a) > 1 and a[0] > a[1]:
    a.reverse()

if len(b) > 1 and b[0] > b[1]:
    b.reverse()

i = 0
j = 0
c = []

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        c.append(a[i])
        i += 1
    else:
        c.append(b[j])
        j += 1

while i < len(a):
    c.append(a[i])
    i += 1

while j < len(b):
    c.append(b[j])
    j += 1

print(*c)
```

W7A22
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất


```python
names = []

n = int(input())

for i in range(n):
    names.append(input())

d = {}

for full_name in names:
    ten = full_name.split()[-1]

    if ten in d:
        d[ten] += 1
    else:
        d[ten] = 1

max_dem = max(d.values())

kq = []

for ten in d:
    if d[ten] == max_dem:
        kq.append(ten)

kq.sort()

print(kq[0])
```

Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi 
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
n = input()
k = int(input())

kq = ""

for i in range(len(n)):
    while len(n) - i + len(kq) > k and kq and kq[-1] < n[i]:
        kq = kq[:-1]

    if len(kq) < k:
        kq += n[i]

print(kq)
```

W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ

```python
s = input()

stack = []

ok = True

for c in s:
    if c in "([{":
        stack.append(c)

    elif c == ")":
        if len(stack) == 0 or stack[-1] != "(":
            ok = False
            break
        stack.pop()

    elif c == "]":
        if len(stack) == 0 or stack[-1] != "[":
            ok = False
            break
        stack.pop()

    elif c == "}":
        if len(stack) == 0 or stack[-1] != "{":
            ok = False
            break
        stack.pop()

if len(stack) != 0:
    ok = False

if ok:
    print("Hop le")
else:
    print("Khong hop le")
```

̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ 
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
n = int(input())

def hop_so(x):
    if x < 4:
        return False

    dem = 0

    for i in range(1, x + 1):
        if x % i == 0:
            dem += 1

    return dem > 2

kq = []
tam = []

for i in range(n + 1, 100):
    if hop_so(i):
        tam.append(i)
    else:
        if len(tam) > len(kq):
            kq = tam
        tam = []

print(*kq)
```

W7A29
Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
a = list(map(int, input().split()))

best = [a[0]]
tam = [a[0]]

for i in range(1, len(a)):
    if a[i] > a[i - 1]:
        tam.append(a[i])
    else:
        if len(tam) > len(best):
            best = tam
        tam = [a[i]]

if len(tam) > len(best):
    best = tam

print(*best)
```

W7A30
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím 

```python
a = list(map(int, input().split()))

n = len(a)

dp = [1] * n
truoc = [-1] * n

for i in range(n):
    for j in range(i):
        if a[j] < a[i]:
            if dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1
                truoc[i] = j

max_i = 0

for i in range(n):
    if dp[i] > dp[max_i]:
        max_i = i

kq = []

while max_i != -1:
    kq.append(a[max_i])
    max_i = truoc[max_i]

kq.reverse()

print(*kq)
```

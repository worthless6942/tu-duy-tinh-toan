# W5A1

Viết hàm trả về số lớn hơn trong hai số nguyên.

```python
def so_lon_hon(a, b):
    if a > b:
        return a
    return b

a = int(input("Nhập a: "))
b = int(input("Nhập b: "))
print(so_lon_hon(a, b))
```

# W5A2

Viết hàm hoán đổi hai số nguyên.

```python
def hoan_doi(a, b):
    return b, a

a = int(input("Nhập a: "))
b = int(input("Nhập b: "))
a, b = hoan_doi(a, b)
print(a, b)
```

# W5A3

Kiểm tra số nguyên tố.

```python
def la_snt(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

n = int(input("Nhập n: "))
print(la_snt(n))
```

# W5A4

Kiểm tra số hoàn hảo.

```python
def la_shh(n):
    tong = 0
    for i in range(1, n):
        if n % i == 0:
            tong += i
    return tong == n

n = int(input("Nhập n: "))
print(la_shh(n))
```

# W5A5

Tìm vị trí đầu tiên của k trong list.

```python
def tim_vi_tri(ds, k):
    for i in range(len(ds)):
        if ds[i] == k:
            return i
    return -1

ds = list(map(int, input("Nhập list: ").split()))
k = int(input("Nhập k: "))
print(tim_vi_tri(ds, k))
```

# W5A6

Tính giai thừa.

```python
def giai_thua(n):
    kq = 1
    for i in range(1, n + 1):
        kq *= i
    return kq

n = int(input("Nhập n: "))
print(giai_thua(n))
```

# W5A7

Máy tính bỏ túi + - * /.

```python
def tinh(a, b, pt):
    if pt == "+":
        return round(a + b, 2)
    if pt == "-":
        return round(a - b, 2)
    if pt == "*":
        return round(a * b, 2)
    if pt == "/":
        return round(a / b, 2)

a = float(input("Nhập số 1: "))
b = float(input("Nhập số 2: "))
pt = input("Nhập phép toán: ")
print(tinh(a, b, pt))
```

# W5A8

Tính khoảng cách Hamming giữa hai số nguyên.

```python
def hamming(x, y):
    return bin(x ^ y).count("1")

x = int(input("Nhập x: "))
y = int(input("Nhập y: "))
print(hamming(x, y))
```

# W5A9

Tổng các chữ số của n.

```python
def tong_chu_so(n):
    tong = 0
    while n > 0:
        tong += n % 10
        n //= 10
    return tong

n = int(input("Nhập n: "))
print(tong_chu_so(n))
```

# W5A10

Kiểm tra hai từ đẳng cấu.

```python
def dang_cau(a, b):
    d1 = {}
    d2 = {}

    for i in range(len(a)):
        if a[i] in d1 and d1[a[i]] != b[i]:
            return False
        if b[i] in d2 and d2[b[i]] != a[i]:
            return False

        d1[a[i]] = b[i]
        d2[b[i]] = a[i]

    return True

a = input("Nhập từ 1: ")
b = input("Nhập từ 2: ")
print(dang_cau(a, b))
```

# W5A12

Chia lấy thương và số dư.

```python
def chia_du(a, b):
    q = a // b
    r = a % b
    return q, r

a = int(input("Nhập số bị chia: "))
b = int(input("Nhập số chia: "))
print(chia_du(a, b))
```

# W5A13

Trả về số lớn hơn trong hai số.

```python
def max_hai_so(a, b):
    return a if a > b else b

a = int(input("Nhập a: "))
b = int(input("Nhập b: "))
print(max_hai_so(a, b))
```

# W5A14

Thực hiện phép tính theo op.

```python
def compute(a, b, op):
    if op == "+":
        return a + b
    if op == "-":
        return a - b
    if op == "*":
        return a * b
    if op == "/":
        return a / b

a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
op = input("Nhập op: ")
print(compute(a, b, op))
```

# W5A15

Kiểm tra tam giác vuông.

```python
def tam_giac_vuong(a, b, c):
    ds = sorted([a, b, c])
    return ds[0] ** 2 + ds[1] ** 2 == ds[2] ** 2

a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))
c = float(input("Nhập cạnh c: "))
print(tam_giac_vuong(a, b, c))
```

```python
def sum_1_to_n(n):
    tong = 0

    for x in range(1, n + 1):
        s = str(x)

        if len(s) == len(set(s)):
            tong += x

    return tong


n = int(input())

print(sum_1_to_n(n))
```

```python
def ucln(a, b):
    ucln = 0
    if a > b:
        a,b = b,a
        
    for i in range (1, b+1):
        if b%i == 0 and a%i == 0 and i > ucln:
            ucln = i
    return ucln

a, b = map(int, input().split())
print(ucln(a,b))
            

```

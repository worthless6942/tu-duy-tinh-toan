# W6A1

Loại bỏ phần tử trùng, giữ lần xuất hiện đầu tiên.

```python
ds = list(map(int, input("Nhập các số: ").split()))

kq = []
for x in ds:
    if x not in kq:
        kq.append(x)

print(kq)
```

# W6A2

```python
ds = list(map(int, input("Nhập các số: ").split()))

tong = 0
kq = []

for x in ds:
    tong += x
    kq.append(tong)

print(kq)
```

# W6A4

Tạo dictionary key -> list value.

```python
du_lieu = input("Nhập các cặp key:value: ").split()

tu_dien = {}

for cap in du_lieu:
    key = cap[:cap.index(":")]
    value = cap[cap.index(":") + 1:]

    if key not in tu_dien:
        tu_dien[key] = []

    tu_dien[key].append(value)

print(tu_dien)
```

# W6A5

Đếm số dương, âm, 0.

```python
ds = list(map(int, input("Nhập list: ").split()))

kq = {
    "positives": 0,
    "negatives": 0,
    "zeros": 0
}

for x in ds:
    if x > 0:
        kq["positives"] += 1
    elif x < 0:
        kq["negatives"] += 1
    else:
        kq["zeros"] += 1

print(kq)
```

# W6A9

Gộp hai dictionary.

```python
d1 = eval(input("Nhập dict 1: "))
d2 = eval(input("Nhập dict 2: "))

kq = {}

for k in d1:
    kq[k] = d1[k]

for k in d2:
    if k in kq:
        kq[k] += d2[k]
    else:
        kq[k] = d2[k]

kq = dict(sorted(kq.items()))
print(kq)
```

# W6A10

Các cặp có tổng bằng k.

```python
ds = list(map(int, input("Nhập list: ").split()))
k = int(input("Nhập k: "))

kq = []

for i in range(len(ds)):
    for j in range(i + 1, len(ds)):
        if ds[i] + ds[j] == k:
            kq.append((ds[i], ds[j]))

kq.sort()
print(kq)
```

# W6A11

Tách tuple chẵn và lẻ.

```python
bo = tuple(map(int, input("Nhập tuple: ").split()))

chan = []
le = []

for x in bo:
    if x % 2 == 0:
        chan.append(x)
    else:
        le.append(x)

print(tuple(chan))
print(tuple(le))
```

# W6A12

Tìm phần tử xuất hiện nhiều nhất.

```python
ds = list(map(int, input("Nhập list: ").split()))

dem = {}

for x in ds:
    dem[x] = dem.get(x, 0) + 1

lon_nhat = max(dem.values())

kq = []
for k, v in dem.items():
    if v == lon_nhat:
        kq.append(k)

print(min(kq))
```

# W6A13

Đảo key và value của dictionary.

```python
d = eval(input("Nhập dictionary: "))

kq = {}

for k, v in d.items():
    kq[v] = k

print(kq)
```

# W6A14

Tìm giao của hai list.

```python
ds1 = list(map(int, input("Nhập list 1: ").split()))
ds2 = list(map(int, input("Nhập list 2: ").split()))

kq = []

for x in ds1:
    if x in ds2 and x not in kq:
        kq.append(x)

print(kq)
```

# W6A15

Lọc dictionary theo value > k.

```python
d = eval(input("Nhập dictionary: "))
k = int(input("Nhập k: "))

kq = {}

for key, value in d.items():
    if value > k:
        kq[key] = value

print(kq)
```

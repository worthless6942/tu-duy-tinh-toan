Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x

```python
def is_armstrong(x):
    n = len(str(abs(x)))  
    temp = abs(x)
    s = 0

    while temp > 0:
        digit = temp % 10
        s += digit ** n
        temp //= 10

    return s == abs(x)

x = int(input("Nhap x"))
print(is_armstrong(x))
```

```python
n = int(input("Nhap n: "))

temp = abs(n)
goc = temp
dao = 0

while temp > 0:
    digit = temp % 10
    dao = dao * 10 + digit
    temp //= 10

if dao == goc:
    print(True)
else:
    print(False)
```

W5A20 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.


```python
def detect_number_type(s):
    try:
        int(s)
        return "int"
    except:
        try:
            float(s)
            return "float"
        except:
            return "neither"

s = input()

print(detect_number_type(s))
```

W5A23 Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.


```python
def quadratic_roots_count(a, b, c):
    delta = b * b - 4 * a * c

    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    else:
        return 0


a, b, c = map(float, input().split())

print(quadratic_roots_count(a, b, c))
```

W5A24 Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    if (x2 - x1) * (y3 - y1) == (x3 - x1) * (y2 - y1):
        return True
    else:
        return False


x1, y1 = map(int, input().split())
x2, y2 = map(int, input().split())
x3, y3 = map(int, input().split())

print(are_collinear(x1, y1, x2, y2, x3, y3))
```

Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)

```python
def point_in_quadrilateral(quad, p):
    x, y = p

    xs = []
    ys = []

    for diem in quad:
        xs.append(diem[0])
        ys.append(diem[1])

    if min(xs) <= x <= max(xs) and min(ys) <= y <= max(ys):
        return True
    else:
        return False


quad = []

for i in range(4):
    x, y = map(int, input().split())
    quad.append((x, y))

px, py = map(int, input().split())
p = (px, py)

print(point_in_quadrilateral(quad, p))
```

W5A27 Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def count_non_zero_digits(lst):
    dem = 0

    for n in lst:
        for ch in str(abs(n)):
            if ch != '0':
                dem += 1

    return dem


lst = list(map(int, input().split()))

print(count_non_zero_digits(lst))
```

W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).


```python
import math

def angle_ABC(x1, y1, x2, y2, x3, y3):
    BAx = x1 - x2
    BAy = y1 - y2

    BCx = x3 - x2
    BCy = y3 - y2

    tich_vo_huong = BAx * BCx + BAy * BCy

    do_dai_BA = math.sqrt(BAx**2 + BAy**2)
    do_dai_BC = math.sqrt(BCx**2 + BCy**2)

    cos_theta = tich_vo_huong / (do_dai_BA * do_dai_BC)

    goc = math.degrees(math.acos(cos_theta))

    return goc


x1, y1 = map(float, input().split())
x2, y2 = map(float, input().split())
x3, y3 = map(float, input().split())

print(angle_ABC(x1, y1, x2, y2, x3, y3))
```

W5A30 Viết hàm tính hệ số nhị thức C(n, k).


```python
def factorial(n):
    t = 1

    for i in range(1, n + 1):
        t *= i

    return t


def C(n, k):
    return factorial(n) // (factorial(k) * factorial(n - k))


n, k = map(int, input().split())

print(C(n, k))
```

Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa,
giá km, phí chờ, phụ phí ban đêm và giảm giá.


```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4:
        mo_cua = 11000
        gia_km = 15000
    else:
        mo_cua = 13000
        gia_km = 17000

    tong = mo_cua + km * gia_km + wait_minutes * 500

    if start_hour >= 22 or start_hour < 6:
        tong *= 1.2

    if km > 20:
        tong *= 0.9

    return tong


car_type = int(input())
start_hour = int(input())
km = float(input())
wait_minutes = int(input())

print(smart_taxi_fare(car_type, start_hour, km, wait_minutes))
```

Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau 


```python
def sum_1_to_n(n):
    tong = 0

    for x in range(1, n + 1):
        s = str(x)

        if len(s) == len(set(s)):
            tong += x

    return tong


n = int(input())

print(sum_1_to_n(n))
```

W3A1 Viết một chương trình nhập vào một số nguyên n và in ra các chữ số đảo ngược nhau. 


```python
n = int(input())

dem = 0
while n > 0:
    digit = n % 10
    dem = dem * 10 + digit
    n //= 10

print(dem)
```

W3A13
[Tính tiền điện]
Nhập vào số kWh điện tiêu thụ. Tính tiền điện theo các bậc sau:
- 0-50 kWh: 1500đ/kWh
- 51-100 kWh: 2000đ/kWh
- 100 kWh: 3000đ/kWh 

```python
kwh = int(input())

if kwh <= 50:
    tien = kwh * 1500
elif kwh <= 100:
    tien = 50 * 1500 + (kwh - 50) * 2000
else:
    tien = 50 * 1500 + 50 * 2000 + (kwh - 100) * 3000

print(tien)
```

W3A16
[Rounding]
Nhập vào một số thực: in ra kết quả làm tròn lên ,tròn xuống, làm tròn tới số nguyên gần nhất (không được sử dụng hàm
sẵn có của pyton). Các kết quả cách nhau một dấu cách 


```python
x = float(input())

if x >= 0:
    tron_xuong = int(x)
    tron_len = int(x) if x == int(x) else int(x) + 1
else:
    tron_xuong = int(x) - 1 if x != int(x) else int(x)
    tron_len = int(x)

phan_le = x - tron_xuong

if phan_le >= 0.5:
    gan_nhat = tron_len
else:
    gan_nhat = tron_xuong

print(tron_len, tron_xuong, gan_nhat)
```

W3A17
Cấp số nhân
Cho 4 số a, b, c, d. Hãy kiểm tra xem 4 số này có thể theo thứ tự tạo thành 1 cấp số nhân với công bội nguyên theo đúng
thứ tự a, b, c, d hay không?
Gợi ý : Tìm công bội (b / a) nếu b chia hết cho a, sau đó lấy b nhân công bội và so sánh vs c, c nhân công bội và so sánh vs
d.

```python
a = int(input())
b = int(input())
c = int(input())
d = int(input())

if a == 0:
    if b == c == d == 0:
        print("YES")
    else:
        print("NO")
else:
    if b % a == 0:
        q = b // a

        if b * q == c and c * q == d:
            print("YES")
        else:
            print("NO")
    else:
        print("NO")
```

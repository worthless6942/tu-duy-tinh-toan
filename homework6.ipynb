Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

W6A17
Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng


```python
n = int(input())

a = []

for i in range(n):
    a.append(list(map(int, input().split())))


for i in range(n):
    if i < len(a[i]):
        print(a[i][i], end=" ")

print()

for i in range(n):
    j = n - 1 - i
    if j < len(a[i]):
        print(a[i][j], end=" ")
```

Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất 

```python
def vi_tri_max(a):
    vt = 0
    for i in range(len(a)):
        if a[i] > a[vt]:
            vt = i
    return vt

a = list(map(int, input().split()))
print(vi_tri_max(a))
```

W6A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def tim_k(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1


a = list(map(int, input().split()))
k = int(input())

print(tim_k(a, k))
```

W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
s = input().split()

d = {}

for i in range(len(s)):
    if s[i] not in d:
        d[s[i]] = []
    d[s[i]].append(i)

for key in d:
    d[key] = tuple(d[key])

print(d)
```

W6A23 VIết hàm xóa đi các số <x có trong mảng


```python
def xoa_nho_hon_x(lst, x):
    kq = []

    for i in lst:
        if i >= x:
            kq.append(i)

    return kq


a = list(map(int, input().split()))
x = int(input())

print(xoa_nho_hon_x(a, x))
```

W6A24
Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
def day_0_ve_dau(lst):
    zero = []
    other = []

    for i in lst:
        if i == 0:
            zero.append(i)
        else:
            other.append(i)

    return zero + other


a = list(map(int, input().split()))
print(day_0_ve_dau(a))
```

W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối


```python
a = list(map(int, input().split()))

tong = 0

for i in range(1, len(a) - 1):
    if a[i] > a[i - 1] and a[i] > a[i + 1]:
        tong += a[i]

print(tong)
```

W6A27
Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). nếu không có in ra -1


```python
a = list(map(int, input().split()))
x = int(input())

check = False

for i in range(len(a)):
    if a[i] < x:
        print(i, end=" ")
        check = True

if check == False:
    print(-1)
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự, hay vô danh (không tăng, không giảm)

```python
a = list(map(int, input().split()))

tang = True
giam = True

for i in range(len(a) - 1):
    if a[i] >= a[i + 1]:
        tang = False

    if a[i] <= a[i + 1]:
        giam = False

if tang:
    print("Day tang")

elif giam:
    print("Day giam")

else:
    print("Vo danh")
```
